```python
X = df.drop(columns=["Diabetes_Status"])
y = df["Diabetes_Status"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("Target classes:", y.unique())
```

```python
# Compare selected glucose-related measurements by target
for col in [
    "Fasting_Blood_Glucose",
    "HbA1c",
    "OGTT",
    "Random_Blood_Glucose"
]:
    if col in X.columns:
        print(f"\n{col}")
        display(
            df.groupby("Diabetes_Status")[col]
            .agg(["mean", "std", "min", "max"])
            .round(2)
        )
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])

print("\nTraining class proportions:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting class proportions:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
```

```python
numeric_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()
```

```python
categorical_features = X_train.select_dtypes(
    include=["object", "str", "bool"]
).columns.tolist()
```

```python
numeric_transformer=StandardScaler()
```

```python
categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)
```

```python

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_features),
        ("Categorical", categorical_transformer, categorical_features)
    ]
)

print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
print("Total features:", len(numeric_features) + len(categorical_features))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Create the complete pipeline
lr_model=Pipeline(steps=[
    ("preprocessor",preprocessor),
    ("model",LogisticRegression(
        max_iter=1000,random_state=42))
])
lr_model.fit(X_train,y_train)
#print("Logistic Regression model trained successfully!")
```

```python
y_pred_lr=lr_model.predict(X_test)

print("First 10 actual labels\n")
print(y_test.iloc[:10].tolist())
print("First 10 predicted labels\n")
print(y_pred_lr[:10].tolist())
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred_lr)

print(f"Test Accuracy: {accuracy:.4f}")
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred_lr,
    labels=["Negative", "Positive"],
    zero_division=0
))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm=confusion_matrix(
     y_test,
    y_pred_lr,
    labels=["Negative", "Positive"]
)
display_labels = ["Negative", "Positive"]

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=display_labels
).plot(cmap="Blues", values_format="d")

plt.title("Logistic Regression — Confusion Matrix")
plt.show()

print("Confusion matrix:")
print(cm)
```

```python
from sklearn.metrics import recall_score, precision_score, f1_score

positive_recall = recall_score(
    y_test, y_pred_lr, pos_label="Positive", zero_division=0
)

positive_precision = precision_score(
    y_test, y_pred_lr, pos_label="Positive", zero_division=0
)

positive_f1 = f1_score(
    y_test, y_pred_lr, pos_label="Positive", zero_division=0
)

print(f"Positive Recall:    {positive_recall:.4f}")
print(f"Positive Precision: {positive_precision:.4f}")
print(f"Positive F1-score:  {positive_f1:.4f}")
```

```python
from sklearn.metrics import roc_auc_score

# Calculate ROC-AUC from predicted probabilities
positive_index = list(lr_model.classes_).index("Positive")
y_probability = lr_model.predict_proba(X_test)[:, positive_index]

auc = roc_auc_score(
    (y_test == "Positive").astype(int),
    y_probability)
print("roc-auc",auc)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

rf_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=200,
        max_depth=None,
        min_samples_split=2,
        random_state=42,
        n_jobs=-1
    ))
])

rf_model.fit(X_train, y_train)

```

```python
y_pred_rf = rf_model.predict(X_test)

print("Actual labels:")
print(y_test.iloc[:10].tolist())

print("\nRandom Forest predictions:")
print(y_pred_rf[:10].tolist())
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

rf_accuracy = accuracy_score(y_test, y_pred_rf)

rf_precision = precision_score(
    y_test, y_pred_rf, pos_label="Positive", zero_division=0
)

rf_recall = recall_score(
    y_test, y_pred_rf, pos_label="Positive", zero_division=0
)

rf_f1 = f1_score(
    y_test, y_pred_rf, pos_label="Positive", zero_division=0
)

rf_positive_index = list(rf_model.classes_).index("Positive")
rf_probabilities = rf_model.predict_proba(X_test)[:, rf_positive_index]

rf_auc = roc_auc_score(
    (y_test == "Positive").astype(int),
    rf_probabilities
)

print(classification_report(
    y_test,
    y_pred_rf,
    labels=["Negative", "Positive"],
    zero_division=0
))

print(f"Accuracy:  {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall:    {rf_recall:.4f}")
print(f"F1-score:  {rf_f1:.4f}")
print(f"ROC-AUC:   {rf_auc:.4f}")
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline

# 1. Decision Tree
dt_model = Pipeline(steps=[
    ("preprocessing", preprocessor),
    ("classifier", DecisionTreeClassifier(
        max_depth=8,
        random_state=42
    ))
])

# 2. K-Nearest Neighbors
knn_model = Pipeline(steps=[
    ("preprocessing", preprocessor),
    ("classifier", KNeighborsClassifier(
        n_neighbors=5
    ))
])

# 3. Support Vector Machine
svm_model = Pipeline(steps=[
    ("preprocessing", preprocessor),
    ("classifier", SVC(
        kernel="rbf",
        
        probability=True,
        random_state=42
    ))
])

# Train all three models
models = {
    "Decision Tree": dt_model,
    "KNN": knn_model,
    "SVM": svm_model
}

for name, model in models.items():
    model.fit(X_train, y_train)
    print(f"{name} trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

new_results = []

for name, model in models.items():
    predictions = model.predict(X_test)

    positive_index = list(model.classes_).index("Positive")
    probabilities = model.predict_proba(X_test)[:, positive_index]

    new_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision_Positive": precision_score(
            y_test, predictions,
            pos_label="Positive",
            zero_division=0
        ),
        "Recall_Positive": recall_score(
            y_test, predictions,
            pos_label="Positive",
            zero_division=0
        ),
        "F1_Positive": f1_score(
            y_test, predictions,
            pos_label="Positive",
            zero_division=0
        ),
        "ROC_AUC": roc_auc_score(
            (y_test == "Positive").astype(int),
            probabilities
        )
    })

day6_results = pd.DataFrame(new_results)

display(
    day6_results.set_index("Model").round(4)
)
```

```python

import pandas as pd

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import make_scorer, precision_score, recall_score, f1_score
```

```python

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_models = {
    "Logistic Regression": lr_model,
    "Random Forest": rf_model,
    "Decision Tree": dt_model,
    "KNN": knn_model,
    "SVM": svm_model
}

scoring_metrics = {
    "accuracy": "accuracy",
    "precision_positive": make_scorer(
        precision_score, pos_label="Positive", zero_division=0
    ),
    "recall_positive": make_scorer(
        recall_score, pos_label="Positive", zero_division=0
    ),
    "f1_positive": make_scorer(
        f1_score, pos_label="Positive", zero_division=0
    ),
    "roc_auc": "roc_auc"
}
```

```python

cv_results = []

for name, model in cv_models.items():
    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring_metrics,
        n_jobs=-1,
        return_train_score=False
    )

    cv_results.append({
        "Model": name,
        "Accuracy": scores["test_accuracy"].mean(),
        "Precision_Positive": scores["test_precision_positive"].mean(),
        "Recall_Positive": scores["test_recall_positive"].mean(),
        "F1_Positive": scores["test_f1_positive"].mean(),
        "ROC_AUC": scores["test_roc_auc"].mean()
    })

cv_comparison = pd.DataFrame(cv_results)

cv_comparison = cv_comparison.sort_values(
    by="ROC_AUC",
    ascending=False
).reset_index(drop=True)

cv_comparison.round(4)
```

```python

print("Model ranking based on cross-validation ROC-AUC:")
display(
    cv_comparison[
        ["Model", "ROC_AUC", "Accuracy", "Precision_Positive",
         "Recall_Positive", "F1_Positive"]
    ].round(4)
)

best_model_name = cv_comparison.iloc[0]["Model"]

print("Highest mean ROC-AUC:", best_model_name)
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import pandas as pd
import matplotlib.pyplot as plt

final_candidates = {
    "Logistic Regression": lr_model,
    "Random Forest": rf_model
}

final_results = []

for name, model in final_candidates.items():
    # Fit using the full training set
    model.fit(X_train, y_train)

    # Evaluate on the untouched test set
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, list(model.classes_).index("Positive")]

    final_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision_Positive": precision_score(
            y_test, y_pred, pos_label="Positive", zero_division=0
        ),
        "Recall_Positive": recall_score(
            y_test, y_pred, pos_label="Positive", zero_division=0
        ),
        "F1_Positive": f1_score(
            y_test, y_pred, pos_label="Positive", zero_division=0
        ),
        "ROC_AUC": roc_auc_score(
            (y_test == "Positive").astype(int), y_prob
        )
    })

    print(f"\n{name} — Classification Report")
    print(classification_report(y_test, y_pred, zero_division=0))

    ConfusionMatrixDisplay.from_predictions(
        y_test,
        y_pred,
        labels=["Negative", "Positive"],
        display_labels=["Negative", "Positive"],
        values_format="d"
    )
    plt.title(f"{name} — Test Confusion Matrix")
    plt.show()

final_comparison = pd.DataFrame(final_results)
display(final_comparison.round(4))
```
